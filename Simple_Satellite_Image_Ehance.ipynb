```python
import cv2 as cv
clip = 'footage/train.mp4'
```

```python
def enhance_image_detail(image_array, method="clahe_unsharp", clip_limit=2.0, tile_grid_size=(8, 8), sigma=1.0, alpha=1.5):

    if len(image_array.shape) == 3:
        img = cv.cvtColor(image_array, cv.COLOR_BGR2GRAY)
    else:
        img = image_array

    if method == "clahe_unsharp":
        clahe = cv.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_grid_size)
        clahe_img = clahe.apply(img)
        blurred = cv.GaussianBlur(clahe_img, (0, 0), sigma)
        enhanced = cv.addWeighted(clahe_img, alpha, blurred, -alpha + 1, 0)
        return enhanced
    else:
        print(f"Unknown method {method}")
        return img
```

```python
cap = cv.VideoCapture(clip)

if not cap.isOpened():
    print("Error: cannot open camera.")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        print("Frame capture failed.")
        break

    enhanced = enhance_image_detail(frame, "clahe_unsharp", clip_limit=2.5, tile_grid_size=(8,8), sigma=1.5, alpha=1.8)

    cv.imshow('Original', frame)
    cv.imshow('Enhanced (CLAHE + Unsharp)', enhanced)
    
    if cv.waitKey(40) & 0xFF == 27:
        break

cap.release()
cv.destroyAllWindows()
```
